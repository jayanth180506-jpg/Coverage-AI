# Feature engineering
Inspect inference-available features. AP identity is represented by a training-only mean, not an ordinal integer; the mean below is fitted only on spatial training cells.

```python
from pathlib import Path
import sys, pandas as pd
from sklearn.model_selection import GroupShuffleSplit
ROOT = Path.cwd() if Path('datasets').exists() else Path.cwd().parent
sys.path.insert(0, str(ROOT/'src'))
from train_models import add_spatial_cells
data = add_spatial_cells(pd.read_csv(ROOT / 'datasets/processed/D5_long.csv'))
tr, _ = next(GroupShuffleSplit(n_splits=1, test_size=.2, random_state=42).split(data, groups=data.cell_id))
ap_means = data.iloc[tr].groupby('bssid').rssi.mean().rename('ap_mean_rssi')
features = data[['x','y','floor_id','building_id','bssid','rssi']].copy().join(ap_means, on='bssid')
display(features.describe(include='all').T)
```

```python
import matplotlib.pyplot as plt
features[['x','y','rssi']].sample(min(5000,len(features)), random_state=42).plot.scatter(x='x', y='y', c='rssi', cmap='RdYlGn', s=5)
plt.title('D5 measured signal by position'); plt.show()
```
